### Question 1: MDP Formulation

State Space (S): The state space will be the set of the numbers of possible rods. This means the state space is actually just the set of integers greater than 0

Action Space (A): The action space account for the addition or removal of any number of rods. This means the action space is the set of all integers with limitations based on how many rods can be removed as we can not go below zero total rods
